```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

# 1.Viewing the data

```python
df1=pd.read_csv("dataset_1.csv",header=None)
df2=pd.read_csv("dataset_2.csv",header=None)

print("First 5 rows of dataset_1:")
print(df1.head())
print("\nFirst 5 rows of dataset_2:")
print(df2.head())
print("Shape of dataset_1:")
print(df1.shape)
print("Shape of dataset_2:")
print(df2.shape)
```

# 2.Data Cleaning

##### Column Header Missing

```python
columns=["Serial_no","MQ2","TEMP","HUMIDITY","DATE_TIME"]
df1.columns=columns 
df2.columns=columns
print("First 5 rows of dataset_1 after renaming columns:")
print(df1.head())
print("\nFirst 5 rows of dataset_2 after renaming columns:")
print(df2.head())
```

##### Checking datatype of each column and if any NULL value if exists

```python

print("Data type of dataset_1:")
print(df1.dtypes)   
print("\nData type of dataset_2:")
print(df2.dtypes)

print("\nMissing values in dataset_1:")
print(missing_values_df1 := df1.isnull().sum())
print("\nMissing values in dataset_2:")
print(missing_values_df2 := df2.isnull().sum())
```

##### Changing datatype of Date_Time

```python
df1["DATE_TIME"] = pd.to_datetime(df1["DATE_TIME"])
df2["DATE_TIME"] = pd.to_datetime(df2["DATE_TIME"])

print("Data type of DATE_TIME column in dataset_1:")
print(df1["DATE_TIME"].dtypes)
print("Data type of DATE_TIME column in dataset_2:")    
print(df2["DATE_TIME"].dtypes)
```

##### Finding-Removing any Duplicate records

```python
duplicate_rows_df1 = df1[df1.duplicated()]
print("Duplicate rows in dataset_1:")
print(duplicate_rows_df1)
duplicate_rows_df2 = df2[df2.duplicated()]
print("\nDuplicate rows in dataset_2:")
print(duplicate_rows_df2)
```

##### BoxPlot Showing Outliers if Any

```python
mq2_df1=df1["MQ2"]
temp_df1 = df1["TEMP"]
humidity_df1=df1["HUMIDITY"]
mq2_df2=df2["MQ2"]  
temp_df2=df2["TEMP"]
humidity_df2=df2["HUMIDITY"]

#calculating quartiles
mq2_df1_q1, mq2_df1_q2, mq2_df1_q3 = mq2_df1.quantile([0.25, 0.5, 0.75])
temp_df1_q1, temp_df1_q2, temp_df1_q3 = temp_df1.quantile([0.25, 0.5, 0.75])
humidity_df1_q1, humidity_df1_q2, humidity_df1_q3 = humidity_df1.quantile([0.25, 0.5, 0.75])

mq2_df2_q1, mq2_df2_q2, mq2_df2_q3 = mq2_df2.quantile([0.25, 0.5, 0.75])
temp_df2_q1, temp_df2_q2, temp_df2_q3 = temp_df2.quantile([0.25, 0.5, 0.75])
humidity_df2_q1, humidity_df2_q2, humidity_df2_q3 = humidity_df2.quantile([0.25, 0.5, 0.75])
#calculating IQR
mq2_iqr_df1 = mq2_df1_q3 - mq2_df1_q1
temp_iqr_df1 = temp_df1_q3 - temp_df1_q1
humidity_iqr_df1 = humidity_df1_q3 - humidity_df1_q1

mq2_iqr_df2 = mq2_df2_q3 - mq2_df2_q1
temp_iqr_df2 = temp_df2_q3 - temp_df2_q1
humidity_iqr_df2 = humidity_df2_q3 - humidity_df2_q1

#calculating lower and upper bounds for outliers
mq2_lower_bound_df1 = mq2_df1_q1 - 1.5 * mq2_iqr_df1
mq2_upper_bound_df1 = mq2_df1_q3 + 1.5 *mq2_iqr_df1 

temp_lower_bound_df1 = temp_df1_q1 - 1.5 * temp_iqr_df1
temp_upper_bound_df1 = temp_df1_q3 + 1.5 * temp_iqr_df1

humidity_lower_bound_df1 = humidity_df1_q1 - 1.5 * humidity_iqr_df1
humidity_upper_bound_df1 = humidity_df1_q3 + 1.5 * humidity_iqr_df1

mq2_lower_bound_df2 = mq2_df2_q1 - 1.5 * mq2_iqr_df2
mq2_upper_bound_df2 = mq2_df2_q3 + 1.5 *mq2_iqr_df2

temp_lower_bound_df2 = temp_df2_q1 - 1.5 * temp_iqr_df2
temp_upper_bound_df2 = temp_df2_q3 + 1.5 * temp_iqr_df2

humidity_lower_bound_df2 = humidity_df2_q1 - 1.5 * humidity_iqr_df2
humidity_upper_bound_df2 = humidity_df2_q3 + 1.5 * humidity_iqr_df2
#outliers
outliers_mq2_df1 = df1[(df1["MQ2"] < mq2_lower_bound_df1) | (df1["MQ2"] > mq2_upper_bound_df1)]
outliers_temp_df1 = df1[(df1["TEMP"] < temp_lower_bound_df1) | (df1["TEMP"] > temp_upper_bound_df1)]
outliers_humidity_df1 = df1[(df1["HUMIDITY"] < humidity_lower_bound_df1) | (df1["HUMIDITY"] > humidity_upper_bound_df1)]

outliers_mq2_df2 = df2[(df2["MQ2"] < mq2_lower_bound_df2) | (df2["MQ2"] > mq2_upper_bound_df2)]
outliers_temp_df2 = df2[(df2["TEMP"] < temp_lower_bound_df2) | (df2["TEMP"] > temp_upper_bound_df2)]
outliers_humidity_df2 = df2[(df2["HUMIDITY"] <  humidity_lower_bound_df2) | (df2["HUMIDITY"] > humidity_upper_bound_df2)]

#plotting MQ2 Boxplot for Dataset1
plt.figure(figsize=(12, 5))

ax = sns.boxplot(
    x=mq2_df1,
    color='lightblue',
    showfliers=True,
    flierprops={
        'marker': 'o',
        'markerfacecolor': 'red',
        'markeredgecolor': 'red',
        'markersize': 7
    }
)

# Draw vertical lines for Q1, median, and Q3
ax.axvline(mq2_df1_q1, color='blue', linestyle='--', label=f'Q1 = {mq2_df1_q1:.2f}')
ax.axvline(mq2_df1_q2, color='green', linestyle='--', label=f'Median = {mq2_df1_q2:.2f}')
ax.axvline(mq2_df1_q3, color='purple', linestyle='--', label=f'Q3 = {mq2_df1_q3:.2f}')

# Label the quartile lines
y = 0.15
ax.text(mq2_df1_q1, y, f'Q1\n{mq2_df1_q1:.2f}', color='blue',
        ha='center', va='bottom')
ax.text(mq2_df1_q2, y, f'Median\n{mq2_df1_q2:.2f}', color='green',
        ha='center', va='bottom')
ax.text(mq2_df1_q3, y, f'Q3\n{mq2_df1_q3:.2f}', color='purple',
        ha='center', va='bottom')

plt.title('MQ2 Sensor Readings: Quartiles and Outliers for Dataset1')
plt.xlabel('MQ2 Reading')
plt.legend(loc='upper right')
plt.tight_layout()
plt.show()

#plotting TEMP Boxplot for Dataset1
plt.figure(figsize=(12, 5))

ax = sns.boxplot(
    x=temp_df1,
    color='lightblue',
    showfliers=True,
    flierprops={
        'marker': 'o',
        'markerfacecolor': 'red',
        'markeredgecolor': 'red',
        'markersize': 7
    }
)

# Draw vertical lines for Q1, median, and Q3
ax.axvline(temp_df1_q1, color='blue', linestyle='--', label=f'Q1 = {temp_df1_q1:.2f}')
ax.axvline(temp_df1_q2, color='green', linestyle='--', label=f'Median = {temp_df1_q2:.2f}')
ax.axvline(temp_df1_q3, color='purple', linestyle='--', label=f'Q3 = {temp_df1_q3:.2f}')

# Label the quartile lines
y = 0.15
ax.text(temp_df1_q1, y, f'Q1\n{temp_df1_q1:.2f}', color='blue',
        ha='center', va='bottom')
ax.text(temp_df1_q2, y, f'Median\n{temp_df1_q2:.2f}', color='green',
        ha='center', va='bottom')
ax.text(temp_df1_q3, y, f'Q3\n{temp_df1_q3:.2f}', color='purple',
        ha='center', va='bottom')

plt.title('TEMPERATURE Sensor Readings: Quartiles and Outliers for Dataset1')
plt.xlabel('TEMPERATURE Reading')
plt.legend(loc='upper right')
plt.tight_layout()
plt.show()

#plotting HUMIDITY Boxplot for Dataset1
plt.figure(figsize=(12, 5))

ax = sns.boxplot(
    x=humidity_df1,
    color='lightblue',
    showfliers=True,
    flierprops={
        'marker': 'o',
        'markerfacecolor': 'red',
        'markeredgecolor': 'red',
        'markersize': 7
    }
)

# Draw vertical lines for Q1, median, and Q3
ax.axvline(humidity_df1_q1, color='blue', linestyle='--', label=f'Q1 = {humidity_df1_q1:.2f}')
ax.axvline(humidity_df1_q2, color='green', linestyle='--', label=f'Median = {humidity_df1_q2:.2f}')
ax.axvline(humidity_df1_q3, color='purple', linestyle='--', label=f'Q3 = {humidity_df1_q3:.2f}')

# Label the quartile lines
y = 0.15
ax.text(humidity_df1_q1, y, f'Q1\n{humidity_df1_q1:.2f}', color='blue',
        ha='center', va='bottom')
ax.text(humidity_df1_q2, y, f'Median\n{humidity_df1_q2:.2f}', color='green',
        ha='center', va='bottom')
ax.text(humidity_df1_q3, y, f'Q3\n{humidity_df1_q3:.2f}', color='purple',
        ha='center', va='bottom')

plt.title('HUMIDITY Sensor Readings: Quartiles and Outliers for Dataset1')
plt.xlabel('HUMIDITY Reading')
plt.legend(loc='upper right')
plt.tight_layout()
plt.show()


#plotting MQ2 Boxplot for Dataset2
plt.figure(figsize=(12, 5))

ax = sns.boxplot(
    x=mq2_df2,
    color='lightblue',
    showfliers=True,
    flierprops={
        'marker': 'o',
        'markerfacecolor': 'red',
        'markeredgecolor': 'red',
        'markersize': 7
    }
)

# Draw vertical lines for Q1, median, and Q3
ax.axvline(mq2_df2_q1, color='blue', linestyle='--', label=f'Q1 = {mq2_df2_q1:.2f}')
ax.axvline(mq2_df2_q2, color='green', linestyle='--', label=f'Median = {mq2_df2_q2:.2f}')
ax.axvline(mq2_df2_q3, color='purple', linestyle='--', label=f'Q3 = {mq2_df2_q3:.2f}')

# Label the quartile lines
y = 0.15
ax.text(mq2_df2_q1, y, f'Q1\n{mq2_df2_q1:.2f}', color='blue',
        ha='center', va='bottom')
ax.text(mq2_df2_q2, y, f'Median\n{mq2_df2_q2:.2f}', color='green',
        ha='center', va='bottom')
ax.text(mq2_df2_q3, y, f'Q3\n{mq2_df2_q3:.2f}', color='purple',
        ha='center', va='bottom')

plt.title('MQ2 Sensor Readings: Quartiles and Outliers for Dataset2')
plt.xlabel('MQ2 Reading')
plt.legend(loc='upper right')
plt.tight_layout()
plt.show()

#plotting TEMP Boxplot for Dataset2
plt.figure(figsize=(12, 5))

ax = sns.boxplot(
    x=temp_df2,
    color='lightblue',
    showfliers=True,
    flierprops={
        'marker': 'o',
        'markerfacecolor': 'red',
        'markeredgecolor': 'red',
        'markersize': 7
    }
)

# Draw vertical lines for Q1, median, and Q3
ax.axvline(temp_df2_q1, color='blue', linestyle='--', label=f'Q1 = {temp_df2_q1:.2f}')
ax.axvline(temp_df2_q2, color='green', linestyle='--', label=f'Median = {temp_df2_q2:.2f}')
ax.axvline(temp_df2_q3, color='purple', linestyle='--', label=f'Q3 = {temp_df2_q3:.2f}')

# Label the quartile lines
y = 0.15
ax.text(temp_df2_q1, y, f'Q1\n{temp_df2_q1:.2f}', color='blue',
        ha='center', va='bottom')
ax.text(temp_df2_q2, y, f'Median\n{temp_df2_q2:.2f}', color='green',
        ha='center', va='bottom')
ax.text(temp_df2_q3, y, f'Q3\n{temp_df2_q3:.2f}', color='purple',
        ha='center', va='bottom')

plt.title('TEMPERATURE Sensor Readings: Quartiles and Outliers for Dataset2')
plt.xlabel('TEMPERATURE Reading')
plt.legend(loc='upper right')
plt.tight_layout()
plt.show()

#plotting HUMIDITY Boxplot for Dataset2
plt.figure(figsize=(12, 5))

ax = sns.boxplot(
    x=humidity_df2,
    color='lightblue',
    showfliers=True,
    flierprops={
        'marker': 'o',
        'markerfacecolor': 'red',
        'markeredgecolor': 'red',
        'markersize': 7
    }
)

# Draw vertical lines for Q1, median, and Q3
ax.axvline(humidity_df2_q1, color='blue', linestyle='--', label=f'Q1 = {humidity_df2_q1:.2f}')
ax.axvline(humidity_df2_q2, color='green', linestyle='--', label=f'Median = {humidity_df2_q2:.2f}')
ax.axvline(humidity_df2_q3, color='purple', linestyle='--', label=f'Q3 = {humidity_df2_q3:.2f}')

# Label the quartile lines
y = 0.15
ax.text(humidity_df2_q1, y, f'Q1\n{humidity_df2_q1:.2f}', color='blue',
        ha='center', va='bottom')
ax.text(humidity_df2_q2, y, f'Median\n{humidity_df2_q2:.2f}', color='green',
        ha='center', va='bottom')
ax.text(humidity_df2_q3, y, f'Q3\n{humidity_df2_q3:.2f}', color='purple',
        ha='center', va='bottom')

plt.title('HUMIDITY Sensor Readings: Quartiles and Outliers for Dataset2')
plt.xlabel('HUMIDITY Reading')
plt.legend(loc='upper right')
plt.tight_layout()
plt.show()


# Listing outliers for all variables in both datasets

outlier_results = {
    "Dataset 1 - MQ2": (outliers_mq2_df1, "MQ2"),
    "Dataset 1 - TEMP": (outliers_temp_df1, "TEMP"),
    "Dataset 1 - HUMIDITY": (outliers_humidity_df1, "HUMIDITY"),
    "Dataset 2 - MQ2": (outliers_mq2_df2, "MQ2"),
    "Dataset 2 - TEMP": (outliers_temp_df2, "TEMP"),
    "Dataset 2 - HUMIDITY": (outliers_humidity_df2, "HUMIDITY")
}

for key, (outlier_df, column) in outlier_results.items():
    print(f"\n{key}")
    print("-" * 40)

    if outlier_df.empty:
        print("No outliers found.")
    else:
        print("Outlier count:", len(outlier_df))
        print("Outlier values:")
        print(outlier_df[column].tolist())
        
```

```python
#Flagging for normal dataset consisting of only normal/warning values
#Clearly if we see all intervals and bounds, we get maximum bound nearest to the data in 99% interval of dataset1 is mq2_upper_bound_df2 hence we consider it as upper fence of our normal condition.

normal_flag_value = mq2_upper_bound_df2
print("normal_flag_value:", round(normal_flag_value, 2))

#similarly we consider the interval of warning as interval between upperbound of dataset 2 and extreme outlier (q3 + 3.0 * iqr) as our warning condition flag value.

warning_flag_value = mq2_df2_q3 + 3.0 * mq2_iqr_df2
print("warning_flag_value:", round(warning_flag_value, 2))
```
